```python
def load_fnirs_data(participant1_file, participant2_file):
# files_dict['s1'] >> path to the s1 snirf
# files_dict['s2'] >> path to the s2 snirf
    """
    Load and preprocess fNIRS data from SNIRF files
    
    Returns:
        tuple: (raw1_hbo, raw2_hbo, raw1_od, raw2_od, events1, events2)
    """
    
    # Load raw data
    raw1 = mne.io.read_raw_snirf(participant1_file, preload=True)
    raw2 = mne.io.read_raw_snirf(participant2_file, preload=True)
    
    # Convert to optical density (needed for SCI)
    raw1_od = mne.preprocessing.nirs.optical_density(raw1)
    raw2_od = mne.preprocessing.nirs.optical_density(raw2)

    # Apply Beer-Lambert law
    raw1_haemo = mne.preprocessing.nirs.beer_lambert_law(raw1_od, ppf=0.1)
    raw2_haemo = mne.preprocessing.nirs.beer_lambert_law(raw2_od, ppf=0.1)
    
    # Extract HbO channels
    raw1_hbo = raw1_haemo.copy().pick_types(fnirs='hbo')
    raw2_hbo = raw2_haemo.copy().pick_types(fnirs='hbo')
    
    # Apply basic filtering to HbO data
    raw1_hbo.filter(0.01, 0.5, verbose=False)
    raw2_hbo.filter(0.01, 0.5, verbose=False)
    
    # Extract events
    events1, _ = mne.events_from_annotations(raw1_hbo, verbose=False)
    events2, _ = mne.events_from_annotations(raw2_hbo, verbose=False)
 
    return raw1_hbo, raw2_hbo, raw1_od, raw2_od, events1, events2


raw1_hbo, raw2_hbo, raw1_od, raw2_od, events1, events2 = load_fnirs_data(files_dict['s1'], files_dict['s2'])
```

```python
class MyPreprocessor(fnirs.BasePreprocessor[mne.io.Raw]):
     
    def read_file(self, path:str, verbose:bool=False):
        if fnirs.DataBrowser.is_path_fif(path):
            return mne.io.read_raw_fif(path, preload=True, verbose=verbose)

        if fnirs.DataBrowser.is_path_nirx(path):
            return mne.io.read_raw_nirx(fname=path, preload=True, verbose=verbose)

        if fnirs.DataBrowser.is_path_snirf(path):
            return mne.io.read_raw_snirf(path, preload=True, verbose=verbose)

        return None
         
    def run(self, raw: mne.io.Raw, verbose: bool = False) -> list[fnirs.MneStep]:
        print('hello')
        raw1_hbo =raw
        # Convert to optical density (needed for SCI)
        raw1_od = mne.preprocessing.nirs.optical_density(raw)

        # SCALP COUPLING INDEX (SCI) - computed on all channels
        sci_p1 = mne.preprocessing.nirs.scalp_coupling_index(raw1_od)

        # Only SCI values for HbO channels
        n_hbo_channels = len(raw1_hbo.ch_names)
        sci_hbo_p1 = sci_p1[:n_hbo_channels] 

        
        # Identify bad channels based on SCI
        bad_channels_p1 = np.where(sci_hbo_p1 < 0.5)[0]  # Use 0.5 threshold
        bad_channel_names_p1 = [raw1_hbo.ch_names[i] for i in bad_channels_p1]
        print(bad_channels_p1)
        print(bad_channel_names_p1)

        # Mark bad channels
        raw1_hbo.info['bads'].extend(bad_channel_names_p1)

        # Apply Beer-Lambert law
        raw1_haemo = mne.preprocessing.nirs.beer_lambert_law(raw1_od, ppf=0.1)
        
        # Extract HbO channels
        raw1_hbo = raw1_haemo.copy().pick_types(fnirs='hbo')
        
        # Apply basic filtering to HbO data
        raw1_hbo.filter(0.01, 0.5, verbose=False)

        
        step = fnirs.MneStep(raw1_hbo, 'hello', 'helllo')
        steps = [step]
        
        return steps
```

```python
def compute_avg_coherence(df, channels, task):

    filtered_df = pd.concat([
        df[
            (df['task'] == task) &
            (df['dyad'] == "s1-s2") &
            (df['channel1'] == df['channel2']) &
            (df['channel1'] == ch)
        ]
        for ch in channels
    ], ignore_index=True)

    avg_coherence = (
        filtered_df.groupby(["channel1", "bin_time_range"], as_index=False)["coherence"]
        .mean()
        .rename(columns={"channel1": "channel", "coherence": "mean_coherence"})
    )

    # ensure bin ranges are sorted numerically
    avg_coherence["bin_start"] = avg_coherence["bin_time_range"].str.split("-").str[0].astype(int)
    avg_coherence = (
        avg_coherence
        .sort_values(["channel", "bin_start"])
        .drop(columns="bin_start")
        .reset_index(drop=True)
    )

    return avg_coherence
```

```python
regions = {
    "TPJ_right": ["S7_D6 hbo", "S7_D8 hbo", "S8_D6 hbo", "S8_D8 hbo"],
    "TPJ_left":  ["S15_D14 hbo", "S15_D15 hbo", "S16_D14 hbo", "S16_D15 hbo"],
    "STS_left":  ["S12_D12 hbo", "S12_D10 hbo", "S12_D11 hbo", "S14_D12 hbo", "S14_D14 hbo", "S14_D11 hbo"],
    "STS_right": ["S4_D4 hbo", "S4_D5 hbo", "S6_D5 hbo", "S6-D4 hbo", "S6_D6 hbo", "S7_D5 hbo"]
}

tasks = ["solo", "collaboration", "free_discussion"]

for region_name, channels in regions.items():
    for task in tasks:
        avg_df = compute_avg_coherence(dataframe, channels, task)
        filename = f"{region_name}_{task}_avg_coherence.csv"
        avg_df.to_csv(filename, index=False)
        print(f"Saved: {filename}")
```
